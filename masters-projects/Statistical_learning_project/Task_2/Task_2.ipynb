```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import numpy as np
```

```python
url = 'https://www.amazon.se/s?k=amazon+apple+macbook+air&language=sv_SE&adgrpid=161053916553&hvadid=692795116979&hvdev=c&hvlocphy=9219559&hvnetw=g&hvqmt=e&hvrand=16145426213559814319&hvtargid=kwd-301976608726&hydadcr=6086_2410891&tag=textstdgledes-21&ref=pd_sl_3fv0x477mc_e'
```

```python
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/130.0.0.0 Safari/537.36"}
```

```python
response = requests.get(url, headers=headers)
```

```python
response
```

```python
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
product_list = soup.find_all('div', attrs = {'class':'a-section a-spacing-small puis-padding-left-small puis-padding-right-small'})
```

```python
product_list
```

```python
product_links = []

for item in product_list:
    for link in item.find_all('a', href=True):
        #print(link['href'])
        product_links.append('https://www.amazon.se' + link['href'])

print(product_links)
```

```python
#test_link = 'https://www.amazon.se/Apple-2022-MacBook-Air-M2-chip/dp/B0B3CZ5MKD/ref=sr_1_5?crid=3FJYIZ56XGSOB&dib=eyJ2IjoiMSJ9.vnRnjF7HQ7EHkoUi_AUkT0WQv0CXkThlSijy-z2OhQDNIfxcYXLWC0_7p9NReuetfZtgL4EJVXFvyn1Yft7JD2SUSnqgOrXWoUAA4nlJlb4xW1D4gtYzDbSmzXfem643tdybSAi_J1guc4XGBACpTIeQVxpk0Du2VynPDzVFeVPOSrjDDXyaIk8Lfrms4kE06Hz_FXJ6jAoMs7TGECzOm2a3-RW1FhlzyEsCPRx44hb2XGnr35I6cApl01XxnMYDHQs8jc0pyVL0UFGZWZGNtVuS73J7erCka5V6YMWK1Vw.aDdGoiLsuuvD2yeObI4y_af5EhLl5YlU3w8WwI88TYE&dib_tag=se&keywords=macbook+air&qid=1729231812&sprefix=%2Caps%2C236&sr=8-5'
```

```python
amazon_products = []

for link in product_links:
    # Skip links that contain 'javascript:void(0)'
    if 'javascript:void(0)' in link:
        print(f"Skipping invalid URL: {link}")
        continue  # Skip the rest of the loop for this URL

    try:
        r = requests.get(link, headers=headers)
        soup = BeautifulSoup(r.content, 'html.parser')
        
        # Extract the product name
        name_tag = soup.find('h1', attrs={'class': 'a-size-large a-spacing-none'})
        name = name_tag.text.strip() if name_tag else "Name not found"
        
        # Extract the product price
        price_tag = soup.find('span', attrs={'class': 'a-price-whole'})
        price = price_tag.text.strip().replace(u'\xa0', u' ').replace(u',', u'') if price_tag else "Price not found"
        
        # Add the product information to the list
        amazon_product = {
            'name': name,
            'price': price
        }
        amazon_products.append(amazon_product)
        print('Saving: ' + amazon_product['name'])
    
    except Exception as e:
        print(f"Error occurred for URL: {link} - {e}")

    

```

```python
amazon_df = pd.DataFrame(amazon_products)
```

```python
print(amazon_df)
```

```python
amazon_df.to_csv("amazon_data.csv", header=True, index=False)
```
