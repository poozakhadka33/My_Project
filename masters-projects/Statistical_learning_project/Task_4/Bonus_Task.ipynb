```python
# Import the necessary libraries
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time
```

```python
# Step 1: Set up WebDriver
# Set up the path to your ChromeDriver
chrome_driver_path = 'C:\\chromedriver-win64\\chromedriver-win64\\chromedriver.exe'

# Create a Service object with the ChromeDriver path
service = Service(executable_path=chrome_driver_path)

# Initialize the WebDriver with the Service object
driver = webdriver.Chrome(service=service)

# Step 2: Open the Nopecha demo page (easy difficulty)
driver.get("https://nopecha.com/captcha/recaptcha#easy")

# Wait for the page to load completely
time.sleep(3)  # Waiting 3 seconds for the page to fully load

# Step 3: Switch to the reCAPTCHA iframe
# reCAPTCHA is often inside an iframe, so we need to switch to that frame to interact with it
frames = driver.find_elements(By.TAG_NAME, "iframe")
for frame in frames:
    if "recaptcha" in frame.get_attribute("src"):
        driver.switch_to.frame(frame)
        break

# Step 4: Locate the "I am not a robot" checkbox and click it
try:
    # Find the checkbox using its class name and click it
    checkbox = driver.find_element(By.CLASS_NAME, "recaptcha-checkbox-border")
    checkbox.click()

    # Step 5: Wait for the CAPTCHA to process (give it a few seconds)
    time.sleep(5)  # This allows the CAPTCHA to verify, which might trigger image challenges

    # After a few seconds, return to the main page (out of iframe)
    driver.switch_to.default_content()

    # Step 6: Check if the CAPTCHA was solved (this varies depending on the implementation)
    print("Attempted CAPTCHA solution.")
    
except Exception as e:
    print(f"Error occurred: {e}")

# Step 7: Close the browser after completing the task
driver.quit()
```
