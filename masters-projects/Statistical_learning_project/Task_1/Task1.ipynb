```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://arxiv.org/html/2410.12790v1'
url2 = 'https://en.wikipedia.org/wiki/Deep_learning'
```

```python
requests.get(url)
```

```python
requests.get(url2)
```

```python
page = requests.get(url)
page2 = requests.get(url2)
```

```python
soup = BeautifulSoup(page.text, 'html')
soup2 = BeautifulSoup(page2.text, 'html')
```

```python
print(soup)
```

```python
print(soup2)
```

```python
headline1 = soup.find_all('h1', class_='ltx_title ltx_title_document')
```

```python
heading1 = soup.find_all('h2', class_='ltx_title ltx_title_section')
```

```python
information1 = soup.find_all('p', class_='ltx_p')
```

```python
file = open("Task_1.txt", "w")
file.write("==================================Scraped Information from Website 1==================================================== \n\n\n")
file.close()
```

```python
out_filename = "Task_1.txt"
file = open(out_filename, "a")
for headl in headline1:
    headl.text
    file.write(headl.text + "\n")
    print(headl.text)
    

for head, info in zip(heading1, information1):
    head.text
    info.text
    file.write(head.text + "\n" + info.text + "\n\n")
    print(head.text+ "\n" +info.text)

file.close()
```

```python
headline2 = soup2.find_all('h1', class_='firstHeading mw-first-heading')
```

```python
information2 = soup2.find_all('p')
```

```python
out_filename = "Task_1.txt"
file = open(out_filename, "a", encoding="utf-8")
for headl in headline2:
    h = headl.text
    file.write(h + "\n")
    print(h)
    

for info in information2:
    p = info.text
    file.write(p + "\n\n")
    print(p)

file.close()
```
